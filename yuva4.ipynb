```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All required libraries imported successfully!")
```

```python
data = {
    "Student": [
        "Amit", "Priya", "Rahul", "Sneha", "Neha",
        "Rohan", "Pooja", "Karan", "Anjali", "Vikas"
    ],
    "Age": [20, 21, 20, 22, 21, 20, 22, 21, 20, 22],
    "Subject": [
        "Python", "Python", "SQL", "SQL", "Python",
        "SQL", "Python", "SQL", "Python", "SQL"
    ],
    "Marks": [85, 92, 78, 88, 95, 72, 89, 80, 91, 75],
    "Study_Hours": [3, 5, 2, 4, 6, 2, 4, 3, 5, 2]
}

df = pd.DataFrame(data)

print("Student Performance Dataset Created Successfully!")
df
```

```python
high_scores = df[df["Marks"] > 85]

print("Students Scoring Above 85:")
high_scores
```

```python
filtered_students = df[
    (df["Marks"] > 85) &
    (df["Subject"] == "Python")
]

filtered_students
```

```python
sorted_df = df.sort_values(
    by="Marks",
    ascending=False
)

sorted_df
```

```python
df.sort_values(by="Marks", ascending=True)
```

```python
subject_count = df.groupby("Subject")["Student"].count()

print("Number of Students in Each Subject:")
print(subject_count)
```

```python
subject_average = df.groupby("Subject")["Marks"].mean()

print("Average Marks by Subject:")
print(subject_average)
```

```python
df["Result"] = np.where(
    df["Marks"] >= 40,
    "Pass",
    "Fail"
)

df
```

```python
marks = df["Marks"].to_numpy()

print("Marks Array:")
print(marks)
```

```python
mean_marks = np.mean(marks)

print("Average Marks:", mean_marks)
```

```python
median_marks = np.median(marks)

print("Median Marks:", median_marks)
```

```python
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
standard_deviation = np.std(marks)

print("Standard Deviation:", round(standard_deviation, 2))
```

```python
total_marks = np.sum(marks)

print("Total Marks:", total_marks)
```

```python
df["Percentage"] = (df["Marks"] / 100) * 100

df[["Student", "Marks", "Percentage"]]
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

```python
df_missing = df.copy()

df_missing.loc[2, "Marks"] = np.nan
df_missing.loc[5, "Study_Hours"] = np.nan

print("Dataset with Missing Values:")
df_missing
```

```python
df_missing["Marks"] = df_missing["Marks"].fillna(
    df_missing["Marks"].median()
)

df_missing["Study_Hours"] = df_missing["Study_Hours"].fillna(
    df_missing["Study_Hours"].median()
)

print("Missing Values After Cleaning:")
print(df_missing.isnull().sum())
```

```python
sns.boxplot(data=df, y="Marks")

plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df["Student"],
    df["Marks"],
    marker="o"
)

plt.title("Student Marks Comparison")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Marks"],
    bins=5,
    edgecolor="black"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Subject",
    y="Marks",
    errorbar=None
)

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Subject",
    y="Marks"
)

plt.title("Marks Distribution by Subject")
plt.xlabel("Subject")
plt.ylabel("Marks")

plt.show()
```

```python
numeric_columns = [
    "Age",
    "Marks",
    "Study_Hours"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")

plt.show()
```
