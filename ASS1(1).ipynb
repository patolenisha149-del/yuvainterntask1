```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
data = { "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108], "Product": [ "Laptop", "Mouse", "Keyboard", "Monitor", "Laptop", "Mouse", "Monitor", "Keyboard" ], "Category": [ "Electronics", "Accessories", "Accessories", "Electronics", "Electronics", "Accessories", "Electronics", "Accessories" ], "Quantity": [2, 10, 5, 4, 1, 8, 3, 6], "Unit_Price": [50000, 800, 1500, 10000, 50000, 800, 10000, 1500] }
```

```python
df = pd.DataFrame(data) 
print(df)
```

```python
print("First Five Rows:") 
print(df.head())
```

```python
print("\nDataset Information:") 
df.info()
```

```python
print("\nDataset Dimensions:") 
print(df.shape)
```

```python
print("\nColumn Names:")
```

```python
print(df.columns.tolist())
```

```python
df["Revenue"] = df["Quantity"] * df["Unit_Price"] 
print(df[["Order_ID", "Product", "Quantity", "Unit_Price", "Revenue"]])
```

```python
total_revenue = df["Revenue"].sum() 
average_revenue = df["Revenue"].mean() 
total_quantity = df["Quantity"].sum()
```

```python
print("Total Revenue:", total_revenue) 
print("Average Revenue per Order:", average_revenue) 
print("Total Quantity Sold:", total_quantity)
```

```python
highest_order = df.loc[df["Revenue"].idxmax()] 
print("Highest-Revenue Order:") 
print(highest_order)
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum() 
print(product_revenue)
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum() 
print(category_revenue)
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum() 
plt.figure(figsize=(8, 5)) 
plt.bar(product_revenue.index, product_revenue.values) 
plt.title("Revenue by Product") 
plt.xlabel("Product") 
plt.ylabel("Revenue") 
plt.tight_layout() 
plt.show()
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum() 
sorted_revenue = product_revenue.sort_values(ascending=False) 
print(sorted_revenue)
```

```python
high_value_orders = df[df["Revenue"] > 20000] 
print(high_value_orders)
```
