```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

print("All required libraries are imported successfully!")
```

```python
business_objectives = {
    "Total Revenue": "Measure total sales income",
    "Total Orders": "Count the number of unique orders",
    "Total Units Sold": "Measure the quantity of products sold",
    "Average Order Value": "Calculate average revenue per order",
    "Top Product": "Identify the highest-revenue product",
    "Best Region": "Identify the highest-revenue region",
    "Monthly Performance": "Track changes in monthly revenue"
}

print("BUSINESS OBJECTIVES")
print("=" * 50)

for objective, description in business_objectives.items():
    print(f"{objective}: {description}")
```

```python
data = {
    "Order_ID": list(range(101, 125)),

    "Order_Date": [
        "2026-01-05", "2026-01-12", "2026-01-18", "2026-01-25",
        "2026-02-04", "2026-02-10", "2026-02-17", "2026-02-24",
        "2026-03-03", "2026-03-09", "2026-03-16", "2026-03-25",
        "2026-04-02", "2026-04-11", "2026-04-19", "2026-04-27",
        "2026-05-05", "2026-05-13", "2026-05-20", "2026-05-28",
        "2026-06-04", "2026-06-12", "2026-06-20", "2026-06-27"
    ],

    "Region": [
        "West", "North", "South", "East",
        "West", "North", "South", "East",
        "West", "North", "South", "East",
        "West", "North", "South", "East",
        "West", "North", "South", "East",
        "West", "North", "South", "East"
    ],

    "Product": [
        "Laptop", "Mobile", "Tablet", "Laptop",
        "Mobile", "Laptop", "Tablet", "Mobile",
        "Laptop", "Tablet", "Mobile", "Laptop",
        "Tablet", "Mobile", "Laptop", "Tablet",
        "Laptop", "Mobile", "Tablet", "Laptop",
        "Mobile", "Tablet", "Laptop", "Mobile"
    ],

    "Units_Sold": [
        10, 20, 15, 8,
        18, 9, 22, 25,
        12, 24, 19, 11,
        20, 28, 10, 26,
        14, 30, 21, 13,
        24, 27, 16, 32
    ],

    "Unit_Price": [
        50000, 15000, 10000, 50000,
        15000, 50000, 10000, 15000,
        50000, 10000, 15000, 50000,
        10000, 15000, 50000, 10000,
        50000, 15000, 10000, 50000,
        15000, 10000, 50000, 15000
    ],

    "Unit_Cost": [
        40000, 11000, 7000, 40000,
        11000, 40000, 7000, 11000,
        40000, 7000, 11000, 40000,
        7000, 11000, 40000, 7000,
        40000, 11000, 7000, 40000,
        11000, 7000, 40000, 11000
    ]
}

df = pd.DataFrame(data)

print(df.head(10))
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Month"] = df["Order_Date"].dt.strftime("%B")

df["Revenue"] = df["Units_Sold"] * df["Unit_Price"]

df["Total_Cost"] = df["Units_Sold"] * df["Unit_Cost"]

df["Profit"] = df["Revenue"] - df["Total_Cost"]

df["Profit_Margin"] = (
    df["Profit"] / df["Revenue"] * 100
)

print(df.head())
```

```python
print("First Five Records:")
print(df.head())

print("\nLast Five Records:")
print(df.tail())

print("\nDataset Dimensions:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Invalid Units Sold:")
print(df[df["Units_Sold"] <= 0])

print("\nInvalid Unit Prices:")
print(df[df["Unit_Price"] <= 0])

print("\nInvalid Unit Costs:")
print(df[df["Unit_Cost"] < 0])

print("\nRecords with Cost Greater Than Selling Price:")
print(df[df["Unit_Cost"] > df["Unit_Price"]])
```

```python
clean_df = df.copy()

clean_df = clean_df.drop_duplicates(
    subset=["Order_ID"]
)

clean_df = clean_df.dropna(
    subset=[
        "Order_ID",
        "Order_Date",
        "Region",
        "Product",
        "Units_Sold",
        "Unit_Price",
        "Unit_Cost"
    ]
)

print("Original Records:", len(df))
print("Records After Cleaning:", len(clean_df))
```

```python
total_revenue = clean_df["Revenue"].sum()

total_cost = clean_df["Total_Cost"].sum()

total_profit = clean_df["Profit"].sum()

total_orders = clean_df["Order_ID"].nunique()

total_units = clean_df["Units_Sold"].sum()

average_order_value = total_revenue / total_orders

overall_profit_margin = (
    total_profit / total_revenue * 100
)

print("=" * 50)
print("KEY PERFORMANCE INDICATORS")
print("=" * 50)

print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Total Cost: ₹{total_cost:,.2f}")
print(f"Total Profit: ₹{total_profit:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Total Units Sold: {total_units}")
print(f"Average Order Value: ₹{average_order_value:,.2f}")
print(f"Overall Profit Margin: {overall_profit_margin:.2f}%")
```

```python
product_analysis = (
    clean_df.groupby("Product")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Units=("Units_Sold", "sum"),
        Total_Cost=("Total_Cost", "sum"),
        Total_Profit=("Profit", "sum"),
        Number_of_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

product_analysis["Profit_Margin"] = (
    product_analysis["Total_Profit"]
    / product_analysis["Total_Revenue"] * 100
)

product_analysis = product_analysis.sort_values(
    by="Total_Revenue",
    ascending=False
)

print(product_analysis)
```

```python
top_product_row = product_analysis.loc[
    product_analysis["Total_Revenue"].idxmax()
]

print("Highest-Revenue Product:", top_product_row["Product"])

print(
    "Revenue: ₹",
    f"{top_product_row['Total_Revenue']:,.2f}"
)

print(
    "Profit: ₹",
    f"{top_product_row['Total_Profit']:,.2f}"
)
```

```python
most_profitable_product = product_analysis.loc[
    product_analysis["Total_Profit"].idxmax()
]

print(
    "Highest-Profit Product:",
    most_profitable_product["Product"]
)

print(
    "Total Profit: ₹",
    f"{most_profitable_product['Total_Profit']:,.2f}"
)
```

```python
region_analysis = (
    clean_df.groupby("Region")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Units=("Units_Sold", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

region_analysis["Revenue_Share_Percent"] = (
    region_analysis["Total_Revenue"]
    / total_revenue * 100
)

region_analysis["Profit_Margin"] = (
    region_analysis["Total_Profit"]
    / region_analysis["Total_Revenue"] * 100
)

region_analysis = region_analysis.sort_values(
    by="Total_Revenue",
    ascending=False
)

print(region_analysis)
```

```python
best_region = region_analysis.loc[
    region_analysis["Total_Revenue"].idxmax()
]

print("Best Region:", best_region["Region"])

print(
    "Regional Revenue: ₹",
    f"{best_region['Total_Revenue']:,.2f}"
)

print(
    "Revenue Contribution:",
    f"{best_region['Revenue_Share_Percent']:.2f}%"
)
```

```python
monthly_analysis = (
    clean_df.groupby(clean_df["Order_Date"].dt.to_period("M"))
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order_ID", "nunique"),
        Total_Units=("Units_Sold", "sum")
    )
    .reset_index()
)

monthly_analysis["Month"] = (
    monthly_analysis["Order_Date"].astype(str)
)

monthly_analysis = monthly_analysis.sort_values(
    by="Order_Date"
)

print(monthly_analysis)
```

```python
monthly_analysis["Revenue_Growth_Percent"] = (
    monthly_analysis["Total_Revenue"].pct_change() * 100
)

print(
    monthly_analysis[
        ["Month", "Total_Revenue", "Revenue_Growth_Percent"]
    ]
)
```
