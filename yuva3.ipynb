```python
df.isnull().sum()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
print(df)
```

```python
df["embarked"] = df["embarked"].fillna( df["embarked"].mode()[0] )
print(df)
```

```python
df.duplicated().sum()
print(df)
```

```python
df= df.drop_duplicates()
print(df)
```

```python
print(df["sex"].unique()) 
print(df["class"].unique()) 
print(df["embarked"].unique())
```

```python
sns.boxplot(x=df["fare"]) 
plt.title("Box Plot of Passenger Fares") 
plt.show()
```

```python
print(df["survived"].value_counts())
```

```python
survival_rate = df["survived"].mean() * 100 
print("Survival Rate:", round(survival_rate, 2), "%")
```

```python
gender_survival = df.groupby("sex")["survived"].mean() * 100 
print(gender_survival)
```

```python
class_survival = df.groupby("class", observed=False)["survived"].mean() * 100 
print(class_survival)
```

```python
print(df["age"].describe())
```

```python
print(df["fare"].describe())
```

```python
sns.countplot(data=df, x="survived") 
plt.title("Titanic Passenger Survival Counts") 
plt.xlabel("Survival Status (0 = No, 1 = Yes)") 
plt.ylabel("Number of Passengers") 
plt.show()
```

```python
sns.barplot(data=df, x="sex", y="survived") 
plt.title("Average Survival Rate by Gender") 
plt.xlabel("Gender") 
plt.ylabel("Survival Rate") 
plt.show()
```

```python
sns.barplot( data=df, x="class", y="survived", order=["First", "Second", "Third"] ) 
plt.title("Survival Rate by Passenger Class") 
plt.xlabel("Passenger Class") 
plt.ylabel("Survival Rate") 
plt.show()
```

```python
sns.histplot(data=df, x="age", bins=20, kde=True) 
plt.title("Age Distribution of Passengers") 
plt.xlabel("Age") 
plt.ylabel("Frequency") 
plt.show()
```

```python
sns.scatterplot( data=df, x="age", y="fare", hue="survived", alpha=0.7 ) 
plt.title("Passenger Age vs Fare by Survival Status") 
plt.xlabel("Age") 
plt.ylabel("Fare") 
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="fare") 
plt.title("Fare Distribution by Passenger Class") 
plt.xlabel("Passenger Class") 
plt.ylabel("Fare") 
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number") 
correlation = numeric_df.corr() 
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f") 
plt.title("Correlation Heatmap") 
plt.show()
```
