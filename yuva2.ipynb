```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
print("Pandas:", pd.__version__) 
print("NumPy:", np.__version__) 
print("Matplotlib:", plt.matplotlib.__version__) 
print("Seaborn:", sns.__version__)
```

```python
data = { "Student": [ "Amit", "Priya", "Rahul", "Sneha", "Neha", "Rohan", "Pooja", "Karan", "Anjali", "Vikas" ], "Age": [20, 21, 20, 22, 21, 20, 22, 21, 20, 22], "Subject": [ "Python", "Python", "SQL", "SQL", "Python", "SQL", "Python", "SQL", "Python", "SQL" ], "Marks": [85, 92, 78, 88, 95, 72, 89, 80, 91, 75], "Study_Hours": [3, 5, 2, 4, 6, 2, 4, 3, 5, 2] }
```

```python
df = pd.DataFrame(data) 
print(df)
```

```python
sorted_df = df.sort_values( by="Marks", ascending=False )
```

```python
print(sorted_df)
```

```python
df["Result"] = np.where( df["Marks"] >= 40, "Pass", "Fail" )
print(df)
```

```python
total_marks = df["Marks"].sum() 
print(total_marks)
```

```python
average_marks = df["Marks"].mean()
print(average_marks)
```

```python
subject_average = df.groupby("Subject")["Marks"].mean() 
print(subject_average)
```

```python
df_missing = df.copy() 
df_missing.loc[2, "Marks"] = np.nan 
print(df_missing)
```

```python
print(df_missing.isnull().sum())
```

```python
mean_marks = df_missing["Marks"].mean() 
df_missing["Marks"] = df_missing["Marks"].fillna(mean_marks) 
print(df_missing)
```

```python
plt.figure(figsize=(10, 5)) 
plt.bar(df["Student"], df["Marks"]) 
plt.title("Student Marks Comparison") 
plt.xlabel("Student Name") 
plt.ylabel("Marks") 
plt.xticks(rotation=45) 
plt.tight_layout() 
plt.show()
```

```python
study_data = df.sort_values("Study_Hours") 
plt.figure(figsize=(8, 5)) 
plt.plot( study_data["Study_Hours"], study_data["Marks"], marker="o" ) 
plt.title("Study Hours and Student Marks") 
plt.xlabel("Study Hours") 
plt.ylabel("Marks") 
plt.grid(True) 
plt.tight_layout() 
plt.show()
```

```python
plt.figure(figsize=(8, 5)) 
sns.scatterplot( data=df, x="Study_Hours", y="Marks", hue="Subject", s=100 ) 
plt.title("Study Hours vs. Student Marks") 
plt.xlabel("Study Hours") 
plt.ylabel("Marks") 
plt.tight_layout() 
plt.show()
```

```python
plt.figure(figsize=(8, 5)) 
plt.hist( df["Marks"], bins=5, edgecolor="black" ) 
plt.title("Distribution of Student Marks") 
plt.xlabel("Marks") 
plt.ylabel("Number of Students") 
plt.tight_layout() 
plt.show()
```

```python
numeric_data = df[ ["Age", "Marks", "Study_Hours"] ] 
correlation_matrix = numeric_data.corr() 
plt.figure(figsize=(7, 5)) 
sns.heatmap( correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f" ) 
plt.title("Correlation Between Numerical Variables") 
plt.tight_layout() 
plt.show()
```
